```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
"""
Data Set Source:
https://raw.githubusercontent.com/keithmcnulty/linear_regression_training/master/advertising.csv
"""

df=pd.read_csv("C:\\Users\\HP\\Downloads\\advertising.csv")
```

```python
# Take TV as X and Sales as Y
X = df["TV"]
Y = df["Sales"]

# Step 1: Calculate mean of X and Y
X_mean = X.mean()
Y_mean = Y.mean()

# Step 2: Calculate slope (m)
# m = Σ((X - X_mean)(Y - Y_mean)) / Σ((X - X_mean)^2)

numerator = 0
denominator = 0

for i in range(len(X)):
    numerator += (X[i] - X_mean) * (Y[i] - Y_mean)
    denominator += (X[i] - X_mean) ** 2

m = numerator / denominator

# Step 3: Calculate intercept (c)
# c = Y_mean - m * X_mean

c = Y_mean - m * X_mean

# Step 4: Calculate predicted Y
Y_pred = []

for x in X:
    prediction = m * x + c
    Y_pred.append(prediction)

# Print results
print("Slope (m):", m)
print("Intercept (c):", c)

print("\nLinear Regression Equation:")
print(f"Sales = {c:.4f} + {m:.4f} * TV")

# Step 5: Plot original data
plt.figure(figsize=(8, 5))

plt.scatter(X, Y)

plt.xlabel("TV")
plt.ylabel("Sales")
plt.title("TV vs Sales - Original Data")

plt.show()

# Step 6: Plot regression line
plt.figure(figsize=(8, 5))

plt.scatter(X, Y, label="Actual Data")
plt.plot(X, Y_pred, label="Regression Line")

plt.xlabel("TV")
plt.ylabel("Sales")
plt.title("TV vs Sales - Linear Regression")

plt.legend()
plt.show()
```
